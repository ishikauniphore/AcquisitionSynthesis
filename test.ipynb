```python
lang = "JA_JP"
```

```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("openai/MMMLU", lang)
df = ds['test'].shuffle().select(range(1000)).to_pandas()
questions = []
answers = []
for i, row in df.iterrows():
    questions.append(f"{row['Question']}\nA: {row['A']}\nB: {row['B']}\nC: {row['C']}\nD: {row['D']}")
    answers.append(row['Answer'])

pd.DataFrame.from_records({
    "questions": questions,
    "answers": answers
}).to_csv(f'data/mmmlu/{lang}.csv')
```

```python
from datasets import load_dataset

ds = load_dataset("Helsinki-NLP/opus-100", f"{code}-en")
df = ds['test'].shuffle().select(range(1000)).to_pandas()
questions = []
answers = []
for row in list(df['translation']):
    questions.append(f"Translate the following from {language} to English: \"{row[code]}\"")
    answers.append(row['en'])

pd.DataFrame.from_records({
    "questions": questions,
    "answers": answers
}).to_csv(f'data/opus-100/{language}.csv')
```
